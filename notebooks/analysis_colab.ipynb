## 📊 Step 5: Үр дүн харах

```python
from IPython.display import Image, display
import json

# Comprehensive visualization харуулах
print("🎨 Comprehensive Analysis Visualization:")
display(Image('results/visualizations/comprehensive_analysis.png'))
```

```python
# Статистик үр дүн харах
print("📈 Statistical Analysis Results:\n")

with open('results/reports/statistical_results.json', 'r', encoding='utf-8') as f:
    stats = json.load(f)

# Descriptive statistics
desc = stats['descriptive_statistics']
print(f"Total Events: {desc['total_events']}")
print(f"Events on Bad Days: {desc['events_on_bad_days']} ({desc['percentage_on_bad_days']:.1f}%)")
print(f"Overall Severity Mean: {desc['severity_overall_mean']:.2f}")

# T-test results
print("\n🧪 T-Test Results:")
t_test = stats['t_test']
print(f"Bad Days Mean: {t_test['bad_day_mean']:.2f}")
print(f"Not Bad Days Mean: {t_test['not_bad_day_mean']:.2f}")
print(f"P-value: {t_test['p_value']:.4f}")
print(f"Significant: {'YES' if t_test['is_significant'] else 'NO'}")

# Overall conclusion
print("\n📋 Overall Conclusion:")
print(stats['overall_conclusion'])
```

## 🔍 Step 7: Нарийвчилсан шинжилгээ (Interactive)

Өгөгдлийг өөрөө судлах

```python
import pandas as pd

# Өгөгдөл унших
matched_df = pd.read_csv('data/processed/matched_events.csv')
lunar_df = pd.read_csv('data/processed/lunar_bad_days_gregorian.csv')
events_df = pd.read_csv('data/processed/world_disasters.csv')

print("✅ Өгөгдөл уншигдлаа!")
print(f"\nТаарсан үйл явдал: {len(matched_df)}")
print(f"Сарны өдрүүд: {len(lunar_df)}")
print(f"Нийт гамшигт үйл явдал: {len(events_df)}")
```

```python
# Муу өдрүүд дээр болсон үйл явдлууд
bad_day_events = matched_df[matched_df['start_is_bad_day'] == True]

print(f"Муу өдрүүд дээр болсон {len(bad_day_events)} үйл явдал:\n")
print(bad_day_events[['event_name', 'event_date', 'severity_score', 'category']].to_string())
```

```python
# Хамгийн их severity бүхий үйл явдлууд
top_events = matched_df.nlargest(10, 'severity_score')

print("Хамгийн их severity бүхий 10 үйл явдал:\n")
print(top_events[['event_name', 'event_date', 'severity_score', 'start_is_bad_day']].to_string())
```
